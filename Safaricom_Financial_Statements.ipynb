### The Income Statement
Consolidated Statement of Profit or Loss.

- **Total Revenue** - The total amount of money generated from the sale of goods or services. It represents the top line of the income statement.
- **Gross Profit** - This is calculated as Total Revenue minus the Cost of Goods Sold (COGS). It indicates how much profit a company makes from its sales after accounting for the direct costs of producing its goods or services.
- **EBITDA** - Earnings Before Interest, Taxes, Depreciation, and Amortization. It's a measure of a company's operating performance, often used to assess a company's profitability without the effects of financing and accounting decisions.
- **Net Profit** - Also known as the bottom line, Net Profit is the amount of money a company has left after all expenses, including operating expenses, interest, and taxes, have been deducted from total revenue.
- **Earnings Per Share (EPS)** - This is a company's Net Profit divided by the number of outstanding shares of its common stock. EPS is a key indicator of a company's profitability and is often used by investors to evaluate a company's financial health.

#### For the year ended 31 March 2025

```python
# Calculating The Margins
revenue        = 388688900000 # KES billions — fill in
gross_profit   = 287607600000 # KES billions — fill in
ebitda         = 172150900000 # KES billions — fill in
net_profit     = 45757200000 # KES billions — fill in

# Margin calculations
gross_margin   = (gross_profit / revenue) * 100
ebitda_margin  = (ebitda / revenue) * 100
net_margin     = (net_profit / revenue) * 100

print("=== SAFARICOM INCOME STATEMENT ANALYSIS ===\n")
print(f"Revenue:         KES {revenue:.1f}B")
print(f"Gross Profit:    KES {gross_profit:.1f}B  ({gross_margin:.1f}% margin)")
print(f"EBITDA:          KES {ebitda:.1f}B  ({ebitda_margin:.1f}% margin)")
print(f"Net Profit:      KES {net_profit:.1f}B  ({net_margin:.1f}% margin)")

print("\n=== INVESTOR INTERPRETATION ===")

if ebitda_margin >= 40:
    print("EBITDA Margin: EXCEPTIONAL - top quartile globally")
elif ebitda_margin >= 30:
    print("EBITDA Margin: STRONG - highly efficient business")
elif ebitda_margin >= 20:
    print("EBITDA Margin: HEALTHY - industry average")
else:
    print("EBITDA Margin: WATCH - below industry average")

if net_margin >= 20:
    print("Net Margin: EXCEPTIONAL - very profitable")
elif net_margin >= 10:
    print("Net Margin: STRONG - good profitability")
elif net_margin >= 5:
    print("Net Margin: MODERATE - acceptable")
else:
    print("Net Margin: THIN - cost pressure concern")
```

### The Balance Sheet

Consolidated Statement of Financial Position in the report.

Looking for six numbers:
- Total Assets - Everything the company owns
- Total Liabilities - Everything it owes
- Total Equity - What belongs to shareholders
- Current Assets - Assets convertible to cash within 1 year
- Current Liabilities - Debts due within 1 year
- Total Debt - Borrowings - bonds, loans, facilities

The golden rule: Assets = Liabilities + Equity.

```python
# Calculating Balance Sheet Ratios
# Fill in from Safaricom's balance sheet
total_assets       = 515284200000   # KES billions
total_liabilities  = 291263100000   # KES billions
total_equity       = 224021100000   # KES billions
current_assets     = 84049200000   # KES billions
current_liabilities= 431235000000   # KES billions
total_debt         = 159393400000  # KES billions

# Key ratios
current_ratio      = current_assets / current_liabilities
debt_to_equity     = total_debt / total_equity
debt_to_ebitda     = total_debt / ebitda  # uses ebitda from above

print("=== SAFARICOM BALANCE SHEET ANALYSIS ===\n")
print(f"Total Assets:      KES {total_assets:.1f}B")
print(f"Total Liabilities: KES {total_liabilities:.1f}B")
print(f"Total Equity:      KES {total_equity:.1f}B")
print(f"\nKey Ratios:")
print(f"Current Ratio:     {current_ratio:.2f}x")
print(f"Debt/Equity:       {debt_to_equity:.2f}x")
print(f"Debt/EBITDA:       {debt_to_ebitda:.2f}x")

print("\n=== INVESTOR INTERPRETATION ===")

if current_ratio >= 1.5:
    print("Liquidity: STRONG - comfortable short-term position")
elif current_ratio >= 1.0:
    print("Liquidity: ADEQUATE - meets short-term obligations")
else:
    print("Liquidity: WATCH - current liabilities exceed current assets")

if debt_to_ebitda <= 2:
    print("Leverage: CONSERVATIVE - low debt burden")
elif debt_to_ebitda <= 4:
    print("Leverage: MODERATE - manageable debt level")
else:
    print("Leverage: HIGH - significant debt relative to earnings")
```

### The Cash Flow Statement
Consolidated Statement of Cash Flows.

most experienced investors read this first.
Three sections to find:
- Operating Cash Flow    - cash from running the business
- Investing Cash Flow    - cash spent on assets, acquisitions
- Financing Cash Flow    - cash from/to shareholders and lenders
                       
**Net Change in Cash**     - did cash go up or down overall?

The most important number: **Free Cash Flow**

*Operating cash flow = Net cash generated from operating activities

*CapEx = Purchase of property and equipment + Acquisition of intangible assets

```python
# Free Cash Flow = Operating Cash Flow - Capital Expenditure
operating_cash_flow = 137693900000   # from cash flow statement
capital_expenditure = 73031200000   # usually negative in investing section

free_cash_flow = operating_cash_flow - abs(capital_expenditure)

fcf_margin = (free_cash_flow / revenue) * 100

print("=== SAFARICOM CASH FLOW ANALYSIS ===\n")
print(f"Operating Cash Flow: KES {operating_cash_flow:.1f}B")
print(f"Capital Expenditure: KES {capital_expenditure:.1f}B")
print(f"Free Cash Flow:      KES {free_cash_flow:.1f}B")
print(f"FCF Margin:          {fcf_margin:.1f}%")

# The quality check — is profit backed by real cash?
cash_conversion = (free_cash_flow / net_profit) * 100
print(f"\nCash Conversion:     {cash_conversion:.1f}%")
print("(>80% = profit is real cash, <50% = accounting concern)")

if cash_conversion >= 80:
    print("Verdict: HIGH QUALITY earnings - profit converts to cash")
elif cash_conversion >= 50:
    print("Verdict: MODERATE quality - some non-cash items")
else:
    print("Verdict: LOW quality - profit not converting to cash")
```

###The Safaricom Paradox - Why 0.19x Current Ratio Isn't A Crisis
A current ratio below 1.0 normally signals danger. But Safaricom is one of the most financially healthy companies in Africa.

How?

Three reasons telecom and fintech companies can safely run low current ratios:
1. **Negative Working Capital Business Model**

Safaricom collects cash from customers before delivering the service. When you buy airtime or deposit M-Pesa float, Safaricom holds your cash immediately. They pay their suppliers later. This means their current liabilities (money owed to suppliers) naturally exceed current assets - not because they're struggling, but because their business model generates cash upfront.

2. **Predictable Recurring Revenue**

A manufacturer with low current ratio is dangerous - if sales slow, they can't pay bills. Safaricom's 40+ million subscribers pay every month like clockwork. Lenders and suppliers know this, so they extend generous payment terms.

3. **Exceptional Operating Cash Flow**

The cash conversion of 141.3% proves the point. Safaricom generates more cash than its reported profit. A company that generates KES X billion in operating cash flow every year doesn't need to hoard current assets as a buffer.

The professional conclusion:

"Safaricom's current ratio of 0.19x initially appears concerning, but is consistent with a negative working capital business model common in telecoms and fintech. The risk is substantially mitigated by exceptional operating cash flow generation, predictable recurring revenue, and conservative leverage of 0.93x Debt/EBITDA. This is not a liquidity risk - it is a structural feature of the business model."
